```python
import pandas as pd

from spending_analyser.loader import load_hsbc_csv
from spending_analyser.categoriser import add_categories

df = add_categories(load_hsbc_csv("../data/sample/hsbc_sample_large.csv"))
spending = df[df["amount"] < 0].copy()
spending["amount"] = spending["amount"].abs()
```

```python
spending["amount"].mean()
```

```python
spending["amount"].median()
```

```python
spending["amount"].describe()
```

```python
q1 = spending["amount"].quantile(0.25)
q3 = spending["amount"].quantile(0.75)
iqr = q3 - q1
upper = q3 + 1.5 * iqr
```

```python
outliers = spending[spending["amount"] > upper]
outliers[["date", "description", "category", "amount"]]
```

```python
def upper_limit(amounts):
    q1 = amounts.quantile(0.25)
    q3 = amounts.quantile(0.75)
    return q3 + 1.5 * (q3 - q1)

spending["limit"] = spending.groupby("category")["amount"].transform(upper_limit)

unusual = spending[spending["amount"] > spending["limit"]]
unusual[["date", "description", "amount", "category", "limit"]]
```

```python
import matplotlib.pyplot as plt

no_rent = spending[spending["category"] != "rent"]
no_rent["amount"].plot(kind="hist", bin=15)
plt.xlabel("$ per transactions")
plt.title("Transactions sizes (excluding rent)")
plt.show()
```
