```python
import sqlite3

from spending_analyser.categoriser import categorise
from spending_analyser.storage import load_transactions

conn = sqlite3.connect("../data/spending.db")
df = load_transactions(conn)

df["rule_says"] = df.apply(lambda row: categorise(row["description"], row["amount"]), axis=1)

coverage = (df["rule_says"] != "Other").mean()

manual = df[df["category_source"] == "manual"]
checked = manual[manual["rule_says"] != "Other"]
accuracy = (checked["rule_says"] == checked["category"]).mean()

print(f"Manual labels: {len(manual)}, of which rules had an answer: {len(checked)}")
print(f"Coverage: {coverage:.0%}")
print(f"Accuracy when rules answer: {accuracy:.0%}")
```
